# **Project Summary -**

This project delivers a comprehensive Exploratory Data Analysis (EDA) on the Google Play Store ecosystem by analyzing two primary datasets: Play Store Data.csv (containing metadata on 9,660 unique apps across 34 categories) and User Reviews.csv (containing over 64,000 translated reviews with sentiment scores). To ensure a structured analytical approach, the research strictly adhered to the UBM Framework—categorizing visual exploration into Univariate, Bivariate, and Multivariate techniques.

The data pipeline began with rigorous preprocessing and cleaning. Outliers in app ratings were removed, categorical numbers in installs and pricing columns were transformed into continuous numeric values, missing entries were handled, and duplicates were eliminated.

The Univariate Analysis (Charts 1–6) examined individual variables to establish baseline market distributions. The findings revealed that Play Store app ratings are heavily left-skewed, maintaining a high overall average of 4.19 out of 5.0. The market is overwhelmingly dominated by free applications, which constitute 92.6% of total listings. Categories like FAMILY, GAME, and TOOLS hold the highest app counts. In terms of sentiment, over 64% of user reviews express positive feedback, with polarity values peaking between 0.0 and 0.4.

The Bivariate Analysis (Charts 7–14) investigated paired relationships between categorical and continuous variables. A critical insight emerged regarding app monetization: free apps achieve exponentially higher installation volumes across all major categories compared to paid alternatives. Furthermore, analysis of pricing trends revealed that paid apps priced between $1.00 and $5.00 maintain strong rating stability, whereas apps priced above $10.00 suffer sharp drops in acquisition—especially within youth-oriented demographics. The analysis also established a strong positive correlation (0.62) between total user reviews and total app installs, proving that user engagement and social proof directly drive app downloads.

The Multivariate Analysis (Charts 15–20 & Pair Plot) evaluated complex interactions across multiple variables simultaneously. Correlation heatmaps and multi-axis scatter plots demonstrated that while high review counts stabilize ratings in the 4.0 to 4.8 range, negative user reviews are predominantly clustered within high subjectivity ranges (0.3 to 0.8). This proves that user dissatisfaction stems primarily from specific operational issues, software bugs, or UI glitches rather than arbitrary dislike.

Ultimately, the project translates raw data into strategic business imperatives, advising stakeholders to adopt Freemium business models, implement targeted in-app review prompts, and utilize sentiment tracking to resolve critical technical bugs.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
````

````python
import pandas as pd
import numpy as np

# 1. Load the dataset
# Assuming 'Play Store Data.csv' is in your working directory
df = pd.read_csv('Play Store Data.csv')
print("--- Initial Info ---")
print(df.info())

# --- Cleaning Steps ---

# 2. Clean 'Installs' (Remove '+' and ',', convert to numeric)
df['Installs'] = df['Installs'].astype(str).str.replace('+', '', regex=False)
df['Installs'] = df['Installs'].astype(str).str.replace(',', '', regex=False)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')

# 3. Clean 'Size' (Convert M to Millions, k to Thousands, handle 'Varies with device')
def clean_size(size_val):
    size_val = str(size_val).strip().lower()
    if 'varies with device' in size_val or 'nan' in size_val:
        return np.nan
    if 'm' in size_val:
        return float(size_val.replace('m', '')) * 1024 * 1024  # or just float(x) * 1e6 depending on preference
    if 'k' in size_val:
        return float(size_val.replace('k', '')) * 1024
    try:
        return float(size_val)
    except ValueError:
        return np.nan

df['Size'] = df['Size'].apply(clean_size)

# 4. Clean 'Price' (Remove '$', convert to float)
df['Price'] = df['Price'].astype(str).str.replace('$', '', regex=False)
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

# 5. Clean 'Reviews' (Convert to integer)
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

# 6. Clean 'Last Updated' (Convert to datetime)
df['Last Updated'] = pd.to_datetime(df['Last Updated'], errors='coerce')

# 7. Handle Missing Values (Imputation strategy)
# Rating: Fill with the median rating of the dataset (or drop if crucial)
df['Rating'] = df['Rating'].fillna(df['Rating'].median())

# Size: Drop rows with missing sizes or fill with group medians
# Here we'll fill with category median so it's more accurate
df['Size'] = df['Size'].groupby(df['Category']).transform(lambda x: x.fillna(x.median()))

# Drop any remaining row anomalies (e.g., if a critical field like 'Reviews' became NaN due to a corrupt row)
df = df.dropna(subset=['Reviews', 'Installs'])

# 8. Reset Index
df = df.reset_index(drop=True)

print("\n--- Cleaned Info ---")
print(df.info())

# Save the analysis-ready data
df.to_csv('Cleaned_Play_Store_Data.csv', index=False)
print("\nDataset successfully cleaned and saved as 'Cleaned_Play_Store_Data.csv'!")
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Style set karna
sns.set_style('whitegrid')
plt.rcParams['font.size'] = 12

# Data load karna
df_apps = pd.read_csv('Play Store Data.csv')
df_reviews = pd.read_csv('User Reviews.csv')

# Data Cleaning
df_apps.drop_duplicates(subset=['App'], inplace=True)
df_apps = df_apps[df_apps['Rating'] <= 5]  # Outlier removal

# Cleaning Installs & Price
df_apps['Installs'] = df_apps['Installs'].astype(str).str.replace('+', '').str.replace(',', '')
df_apps = df_apps[df_apps['Installs'] != 'Free']
df_apps['Installs'] = pd.to_numeric(df_apps['Installs'])

df_apps['Price'] = df_apps['Price'].astype(str).str.replace('$', '')
df_apps['Price'] = pd.to_numeric(df_apps['Price'])

df_apps['Reviews'] = pd.to_numeric(df_apps['Reviews'])
````

#### What do you suggest the client to achieve Business Objective ?
Explain Briefly.

o achieve the primary Business Objective of maximizing downloads, improving user retention, and optimizing revenue generation on the Google Play Store, here are the strategic suggestions for the client based on data insights:

Data Insight: Negative reviews exhibit high subjectivity (between 0.3 and 0.8), indicating that negative sentiment stems from specific technical bugs, poor UI, or feature crashes rather than general dislike.

Recommendation:
